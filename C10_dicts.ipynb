## 1. Dictionaries: key → value
**Dictionary:** pairs of `key: value` in curly brackets `{ }`; you look values up by key, not by position.

```python
student = {"name": "Ayesha", "city": "Lahore", "marks": 88}
print(student["name"])
print(student.get("phone", "not given"))
student["marks"] = 92
student["course"] = "AI"
print(student)
```

```python
print(student.keys())
print("city" in student)
student.pop("city")
print(student)
```

## 2. Loop over a dictionary
**.items():** gives every (key, value) pair, so `for item, qty in stock.items()` gets both.

```python
stock = {"Sugar": 40, "Tea": 8, "Oil": 5}
for item, qty in stock.items():
    if qty <= 7:
        print("Low stock:", item, qty)
print("Total units:", sum(stock.values()))
```

```python
for item in stock:
    print(item, "→", stock[item])
```

```python
votes = ["Ali", "Sara", "Ali", "Hamza", "Ali", "Sara"]
tally = {}
for name in votes:
    tally[name] = tally.get(name, 0) + 1
print(tally)
```

## 3. Dictionary comprehension
**Dict comprehension:** `{key: value for ... in ...}` makes a new dictionary in one line.

```python
prices = {"Sugar": 150, "Tea": 390, "Oil": 560}
sale = {item: round(p * 0.9) for item, p in prices.items()}
print(sale)
```

```python
squares = {n: n * n for n in range(1, 4)}
print(squares)
```

## 4. Nested data: real records
**Nested data:** a list of dictionaries (rows) or a dictionary of lists (columns) — how real data (JSON) looks.

```python
orders = [
    {"item": "Sugar", "qty": 2, "price": 150},
    {"item": "Tea", "qty": 1, "price": 390},
]
bill = 0
for o in orders:
    bill += o["qty"] * o["price"]
print("Bill: Rs.", bill)
```

```python
print(orders[1]["item"])
print(len(orders))
```

```python
marks = {"Ali": [70, 80, 90], "Sara": [88, 92, 95]}
for name, scores in marks.items():
    avg = sum(scores) / len(scores)
    print(name, round(avg, 1))
```

## 5. Sets: unique values
**Set:** an unordered collection of unique values in `{ }`; duplicates disappear.

```python
cities = ["Lahore", "Okara", "Lahore", "Multan", "Okara"]
unique = set(cities)
print(len(unique))
print(sorted(unique))
unique.add("Karachi")
print("Karachi" in unique)
```

```python
letters = set("banana")
print(sorted(letters))
```

```python
python_class = {101, 102, 103, 104}
ai_class = {103, 104, 105}
print(sorted(python_class & ai_class))
print(sorted(python_class | ai_class))
print(sorted(python_class - ai_class))
```

## 6. Choosing the right structure
**Choose:** list for order, tuple for a fixed record, dictionary for lookup by key, set for unique values.
