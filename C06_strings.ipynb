## 1. A string is a row of letters
**Index:** the position of a character; the first one is `0`, and `-1` is the last one.

```python
city = "Lahore"
print(city[0])
print(city[3])
print(city[-1])
print(len(city))
```

```python
word = "Python"
print(word[1], word[-2])
```

```python
print(type("Lahore"))
```

## 2. Slicing: cut out a part
**Slicing:** `text[start:stop:step]` gives a part of a string; `stop` is not included.

```python
date = "14-11-2026"
day = date[0:2]
month = date[3:5]
year = date[6:]
print(day, month, year)
```

```python
phone = "0300-1234567"
print(phone[:4])     # the network code
```

```python
name = "Ramzan"
print(name[:3])
print(name[-3:])
print(name[::-1])
```

```python
print("Muslim Revive Skills"[::2])   # every second letter
```

## 3. Looping through a string
**for ch in text:** the loop gets one character at a time.

```python
text = "banana"
count = 0
for ch in text:
    if ch == "a":
        count += 1
print("a appears", count, "times")
```

```python
print("a" in "banana")
print("z" in "banana")
```

```python
for i in range(len("MRS")):
    print(i, "MRS"[i])
```

## 4. String methods
**Method:** a built-in action of a string, written `text.method()`; it returns a new string.

```python
name = "  muhammad ali  "
clean = name.strip().title()
print(clean)
print(clean.upper())
print(len(name), len(clean))
```

```python
msg = "Python is easy"
print(msg.lower())
print(msg.replace("easy", "fun"))
```

```python
print("2026".isdigit(), "20a6".isdigit())
```

```python
sentence = "Python is easy and Python is fun"
words = sentence.split()
print(words)
print(sentence.count("Python"))
print(sentence.find("easy"))
print("-".join(["14", "11", "2026"]))
```

```python
print("Bismillah, let us start".startswith("Bismillah"))
```

## 5. f-strings: neat output
**f-string:** `f"...{value}..."` puts values into text; `:.2f`, `:,` and `:<10` format them.

```python
item = "Sugar"
qty = 2
price = 150
print(f"{item} x {qty} = Rs. {qty * price}")
print(f"Average: {7 / 3:.2f}")
print(f"Big number: {1234567:,}")
print(f"|{item:<10}|{qty:>5}|")
```

```python
marks = 437
total = 500
print(f"Percentage: {marks / total * 100:.1f}%")
```

## 6. A first look at regular expressions
**Regular expression (regex):** a pattern that describes what to search for in text.

```python
import re
text = "Order 12 bags of rice and 5 kg sugar for Rs. 3450"
print(re.findall(r"\d+", text))
print(re.findall(r"[A-Z]\w+", text))
print(bool(re.fullmatch(r"\d{5}", "54000")))
```

```python
import re
print(re.findall(r"\d+", "Roll 23 got 87 marks"))
```
