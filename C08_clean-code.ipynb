A function takes parameters and returns a value.

## 1. Recursion: a function that calls itself
**Recursion:** a function that calls itself on a smaller problem; the **base case** stops it.

```python
def countdown(n):
    if n == 0:
        print("Go!")
        return
    print(n)
    countdown(n - 1)

countdown(3)
```

```python
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

print(power(2, 4))
```

```python
print(power(3, 3))
print(power(10, 0))
```

## 2. lambda: tiny one-line functions
**lambda:** a small one-line function without a name: `lambda x: x * 2`.

```python
double = lambda x: x * 2
add = lambda a, b: a + b
print(double(7))
print(add(3, 4))
```

```python
a = [1, 2, 4, 3, 5]
b = {"one": 1, "two": 2, "three": 3}
print(a)
print(type(a))
print(b)
print(type(b))
```

```python
is_even = lambda n: n % 2 == 0
print(is_even(8), is_even(7))
```

```python
items = [("Tea", 390), ("Sugar", 150), ("Oil", 560)]
cheap_first = sorted(items, key=lambda item: item[1])
print(cheap_first)
```

```python
names = ["Ayesha", "Ali", "Hamza"]
print(max(names, key=lambda n: len(n)))
```

## 3. Docstrings: explain your function
**Docstring:** the text in triple quotes on the first line of a function, saying what it does.

```python
def area_of_room(length, width):
    """Return the area of a room in square feet."""
    return length * width

print(area_of_room(12, 10))
print(area_of_room.__doc__)
```

```python
help(area_of_room)
```

## 4. PEP 8: code that others can read
**PEP 8:** Python’s official style guide — names, spaces and indentation that make code readable.

```python
def calculate_total(price, quantity, discount=0):
    """Return the bill after the discount."""
    subtotal = price * quantity
    return subtotal - discount

print(calculate_total(150, 3, discount=50))
```

## 5. Errors and bugs
**Bug:** a mistake in code. Python shows most errors; a **logic error** gives a wrong answer without any message.

```python
def count_even(n):
    count = 0
    for i in range(1, n):
        if i % 2 == 0:
            count += 1
    return count

print(count_even(10))
```

```python
# print() debugging: see i and count in every round
def count_even_debug(n):
    count = 0
    for i in range(1, n):
        if i % 2 == 0:
            count += 1
        print("i =", i, "count =", count)
    return count

count_even_debug(10)
```

## 6. The VS Code debugger
**Debugger:** a tool that pauses your code at a breakpoint and runs it one line at a time while you watch the variables.

```python
def count_even(n):
    count = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            count += 1
    return count

print(count_even(10))
```
