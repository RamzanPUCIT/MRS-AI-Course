## 1. Repeat with a for loop
**for loop:** runs the indented block once for every value it gets.

```python
# without a loop: the same line again and again
print("Bismillah")
print("Bismillah")
print("Bismillah")
```

```python
# with a loop: one line, repeated 3 times
for i in range(3):
    print("Bismillah")
```

```python
for lap in range(1, 6):
    print("Lap", lap)
print("Race finished!")
```

```python
for letter in "MRS":      # a for loop can walk through text too
    print(letter)
```

## 2. Counting with range()
**range(start, stop, step):** makes the numbers for a loop; `stop` is never included.

```python
price_per_kg = 150
for kg in range(1, 6):
    print(f"{kg} kg sugar = Rs. {kg * price_per_kg}")
```

```python
for n in range(10, 0, -2):
    print(n, end=" ")
```

```python
for n in range(5):
    print(n, end=" ")
```

```python
for n in range(2, 6):
    print(n, end=" ")
```

```python
for n in range(0, 21, 5):
    print(n, end=" ")
```

## 3. Running totals: the accumulator pattern
**Accumulator:** start a variable at 0 before the loop, update it inside, print it after.

```python
total = 0
for item in range(1, 4):
    price = int(input(f"Price of item {item}: "))
    total = total + price
print("Total bill: Rs.", total)
```

```python
total = 0
for n in range(1, 11):
    total += n
print("1 + 2 + ... + 10 =", total)
```

```python
count = 0
for n in range(1, 21):
    if n % 3 == 0:
        count += 1
print("Multiples of 3 up to 20:", count)
```

## 4. while: repeat until something changes
**while loop:** keeps running as long as its condition is `True`.

```python
savings = 0
week = 0
while savings < 3000:
    savings = savings + 500
    week = week + 1
    print(f"Week {week}: Rs. {savings}")
print("Goal reached in", week, "weeks")
```

```python
n = 5
while n > 0:
    print(n)
    n = n - 1
print("Go!")
```

## 5. break and continue
**continue** skips to the next round; **break** leaves the loop completely.

```python
for n in range(1, 7):
    if n == 3:
        continue
    print(n, end=" ")
```

```python
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(n, end=" ")
```

```python
for attempt in range(1, 4):
    password = input("Password: ")
    if password == "mrs123":
        print("Welcome to MRS!")
        break
    print("Wrong password, try again")
```

```python
n = 1
while True:
    if n * n > 50:
        break
    n += 1
print("First square above 50:", n * n)
```

## 6. Nested loops and patterns
**Nested loop:** a loop inside a loop; the inner loop runs completely for every round of the outer loop.

```python
for row in range(1, 4):
    for col in range(1, 6):
        print("*", end=" ")
    print()
```

```python
for row in range(1, 4):
    print("*" * 5)     # shortcut: repeat a string
```

```python
for a in range(1, 4):
    for b in range(1, 4):
        print(f"{a}x{b}={a*b}", end="   ")
    print()
```
