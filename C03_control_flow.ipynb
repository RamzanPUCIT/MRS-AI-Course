## 1. Making decisions with if
**if:** runs a block of code only when a condition is `True`.

```python
is_raining = False
if is_raining:
    print("It's raining, take an umbrella!")
```

```python
is_raining = True
if is_raining:
    print("Take an umbrella")
print("Have a good day!")
```

## 2. Two paths: if … else
**if … else:** one block runs when the condition is `True`, the other when it is `False`.

```python
number = int(input("Enter a number: "))
```

```python
print(number)
```

```python
if number % 2 == 0:
    print(number, "is even")
```

```python
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
```

## 3. Many choices: elif
**elif:** checks the next condition only if all the ones above were `False`.

```python
marks = 30
```

```python
print(marks)
```

```python
marks = int(input("Enter marks (0-100): "))
if marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"
print("Your grade is", grade)
```

## 5. Problem solving: plan, draw, code, test
**Problem solving:** pseudocode → flowchart → code → dry run.

```python
bill = float(input("Bill amount (Rs.): "))
if bill >= 5000:
    discount = bill * 10 / 100
else:
    discount = 0
print(f"Discount: Rs. {discount}")
print(f"Pay: Rs. {bill - discount}")
```
