## 1. Write once, use many times
**Function:** a named block of code; `def` writes it, and calling its name runs it.

```python
def greet():
    print("Assalam-o-Alaikum!")
    print("Welcome to MRS")

greet()
greet()
```

```python
def line():
    print("-" * 20)

line()
print("MRS")
line()
```

## 2. Parameters and arguments
**Parameter:** a variable in the `def` line; **argument:** the value you send when you call.

```python
def welcome(name, city):
    print(f"{name} from {city}, welcome to MRS!")

welcome("Ayesha", "Lahore")
welcome("Bilal", "Okara")
```

```python
welcome("Hamza", "Multan")
```

## 3. return: get a result back
**return:** sends a value back to the line that called the function, and ends the function.

```python
def bill(qty, price):
    total = qty * price
    return total

sugar = bill(2, 150)
tea = bill(1, 390)
print("Total:", sugar + tea)
```

```python
print(bill(3, 320))
```

```python
def is_pass(marks):
    return marks >= 50

print(is_pass(72), is_pass(41))
```

```python
def show_double(n):
    print(n * 2)
    return n * 2
result = show_double(5)
print(result)
```

## 4. Default and keyword arguments
**Default argument:** `qty=1` is used when the caller leaves it out; **keyword argument:** `unit="litre"` sends a value by name.

```python
def order(item, qty=1, unit="kg"):
    return f"{qty} {unit} {item}"

print(order("Sugar"))
print(order("Rice", 5))
print(order("Milk", unit="litre", qty=2))
```

```python
print(order("Tea", qty=3, unit="packet"))
```

## 5. Any number of values: *args and **kwargs
**\*args** collects extra positional values in a tuple; **\*\*kwargs** collects extra name=value pairs in a dictionary.

```python
def receipt(shop, *items):
    print(f"--- {shop} ---")
    for item in items:
        print("•", item)
    print(f"{len(items)} items")

receipt("MRS Store", "Sugar", "Tea", "Oil")
```

```python
receipt("Corner Shop", "Atta", "Daal")
```

```python
def profile(name, **details):
    print("Name:", name)
    for key, value in details.items():
        print(f"{key}: {value}")

profile("Ayesha", city="Lahore", course="AI")
```

## 6. Scope: where a variable lives
**Scope:** a variable made inside a function is local — it exists only inside that function.

```python
# city = "Lahore"

def change_city():
    city = "Karachi"
    print("Inside:", city)
    # return city
change_city()
print("Outside:", city)
```

```python
def make_total():
    total = 500
    return total

print(make_total())
# print(total)   # remove # to see the NameError: total is local
```
