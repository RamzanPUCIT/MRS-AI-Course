## 1. Variables

```python
ramzan = 2346737434893478437897894378934578934578945789238934
ali = 785747845782983934734778959893893489
c = ramzan + ali
print(c)
```

## 2. Data types
**Data type:** the kind of value — `int`, `float`, `str`, `bool`. Check it with `type()`.

```python
age = 20
type(age)
```

```python
age = 20          # int   -> whole number
price = 99.5      # float -> decimal number
city = "Lahore"   # str   -> text (always inside quotes)
is_open = True    # bool  -> True or False
print(type(age), type(price))
print(type(city), type(is_open))
```

```python
x = 7
print(x, type(x))
x = 7.5
print(x, type(x))
x = "seven"
print(x, type(x))     # the type follows the value
```

## 3. Operators
**Operators:** symbols that calculate (`+ - * / // % **`) or compare (`> == !=`).

```python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b, a // b, a % b, a ** 2)
print(a > b, a == b, a != b)
```

```python
sweets = 17
friends = 5
print("Each friend gets:", sweets // friends)
print("Left over:", sweets % friends)
```

```python
print(2 + 3 * 4)       # * first  -> 14
print((2 + 3) * 4)     # () first -> 20
```

```python
score = 10
score += 5             # same as: score = score + 5
print(score)
score -= 3             # same as: score = score - 3
print(score)
```

```python
print(5 > 3, 5 < 3, 5 == 5, 5 != 5)
```

```python
first = "Muslim "
last = "Revive Skills"
print(first + last)    # + joins text
print("MRS" * 5)        # * repeats text
```

## 4. Input and type conversion
**`input()`** reads text from the keyboard; **`int()` / `float()`** turn that text into a number.

```python
a = input("Enter Your First number: ")  
b = input("Enter Your Second number: ")
print(a + b)              # text is repeated!
print(int(a) + int(b))   # converted to numbers
```

```python
qty = input("Quantity: ")   # the user types 3
print(qty * 2)              # text is repeated!
print(int(qty) * 2)         # converted to a number
```

```python
print(int("25") + 5)
print(float("2.5") + 1)
print(str(100) + " marks")
print(int(9.99))       # int() removes the decimal part
```

```python
name = input("Your name: ")
print("Welcome,", name)
```

```python
num = int(input("Enter a number: "))
print("Double:", num * 2)
```
