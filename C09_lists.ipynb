## 1. Lists: many values in one variable
**List:** an ordered collection of values in square brackets `[ ]`; you can change it.

```python
cart = ["Sugar", "Tea", "Rice"]
print(cart[0])
print(cart[-1])
cart[1] = "Green Tea"
print(cart)
print(len(cart))
```

```python
print(cart[0:2])
print("Rice" in cart)
```

## 2. Add and remove items
**List methods:** `append`, `insert`, `remove` and `pop` change the list itself.

```python
cart = ["Sugar", "Tea"]
cart.append("Oil")
cart.insert(0, "Milk")
cart.remove("Tea")
last = cart.pop()
print(cart)
print("Removed:", last)
```

```python
nums = [5, 2, 9, 1]
nums.sort()
print(nums)
print(nums.index(9), nums.count(2))
```

## 3. Loop over a list
**Loop + built-ins:** `for p in prices` visits every item; `sum`, `max`, `min` and `sorted` work on the whole list.

```python
prices = [150, 390, 560, 220, 90]
count = 0
for p in prices:
    if p > 300:
        count += 1
print("Above 300:", count)
print("Total:", sum(prices))
print("Max:", max(prices), "Min:", min(prices))
```

```python
print(sorted(prices))
print(sorted(prices, reverse=True))
```

## 4. Same list or a copy?
**Alias vs copy:** `b = a` gives the same list a second name; `a.copy()` makes a new list.

```python
a = ["Lahore", "Okara"]
b = a
c = a.copy()
a.append("Karachi")
print(b)
print(c)
```

## 5. List comprehensions
**List comprehension:** `[expression for item in list if condition]` makes a new list in one line.

```python
prices = [150, 390, 560, 220]
with_tax = [round(p * 1.18) for p in prices]
cheap = [p for p in prices if p < 400]
print(with_tax)
print(cheap)
```

```python
names = ["ali", "sara", "hamza"]
print([n.title() for n in names])
```

## 6. Tuples and nested lists
**Tuple:** like a list in round brackets `( )`, but it cannot change. **Nested list:** a list inside a list.

```python
product = ("Sugar", 150, "kg")
name, price, unit = product
print(name, price, unit)
print(product[1])
# product[1] = 160  → TypeError: a tuple cannot change
```

```python
point = (3, 4)
x, y = point
print(x, y)
```

```python
items = [("Tea", 390), ("Sugar", 150), ("Oil", 560)]
total = 0
for name, price in items:
    print(f"{name:<6} Rs. {price}")
    total += price
print("Total: Rs.", total)
```

```python
shelves = [["Tea", "Sugar"], ["Rice", "Daal", "Oil"]]
print(shelves[1][2])
for shelf in shelves:
    print(len(shelf), "items:", shelf)
```

```python
print(len(shelves))
print(shelves[0][-1])
```
